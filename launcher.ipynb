```python
# Cell 4 — download & preview hasil langsung di Kaggle (lebih cepat dari Gradio)
import glob, os, shutil, sys

runs = sorted(glob.glob("repo/outputs/run_*"), key=os.path.getmtime)
assert runs, "Belum ada hasil — jalankan Generate dulu di UI Gradio (Cell 3)."
run = runs[-1]
print(f"Run terbaru: {run}")

sys.path.insert(0, "repo")
from core.export import zip_run
zip_path = zip_run(run)                                   # final.wav + chunks + manifest.json
dest = shutil.move(zip_path, os.path.basename(zip_path))  # → /kaggle/working (panel Files)
print(f"✅ ZIP siap: {dest} ({os.path.getsize(dest)/1e6:.1f} MB)")
print("→ Download: panel Files kanan → /kaggle/working → klik ⋮ pada ZIP → Download")

# Preview 30 detik pertama langsung di notebook (audio penuh ada di ZIP)
import soundfile as sf
from IPython.display import Audio, display
info = sf.info(os.path.join(run, "final.wav"))
wav, sr = sf.read(os.path.join(run, "final.wav"),
                  frames=int(info.samplerate * 30), dtype="float32", always_2d=False)
display(Audio(wav, rate=sr))
```
